```python
df1=pd.read_csv('stock_data.csv')
df1
```

```python
df1['Date']=pd.to_datetime(df['Date'],format = '%d-%m-%Y')
df1.sort_values('Date')
```

```python
stock_column = "Stock_1"
```

```python
df1['Lag_1'] = df1[stock_column].shift(1)
df1['Lag_2'] = df1[stock_column].shift(2)
df1['Lag_3'] = df1[stock_column].shift(3)
```

```python
df1['Lag_3']
```

```python
df1=df1.dropna()
df1
```

```python
X = df1[['Lag_1', 'Lag_2', 'Lag_3']]
y = df1[stock_column]
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.2,shuffle = False)
```

```python
# model training
```

```python
from sklearn.linear_model import LinearRegression
Model = LinearRegression()
```

```python
Model.fit(x_train,y_train)
```

```python
y_pred = Model.predict(x_test)
print((y_test,y_pred))
```

```python
from sklearn.metrics import r2_score,mean_squared_error
```

```python
# Model evaluation
r2_sco = r2_score(y_pred,y_test)
r2_sco
```

```python
mse = mean_squared_error(y_pred,y_test)
mse
```

```python
rmse = np.sqrt(mse)
rmse
```

```python
print(f"\nModel performance")
print("--------------------------")
print(f"Mean squared error: {mse}")
print(f"Root mean squared error: {rmse}")
print(f"r2 score: {r2_sco}")
```

```python
# Visualization
```

```python
plt.figure
plt.plot(y_test.values)
plt.plot(y_pred)
plt.title(f"Actual vs predicted prices {stock_column}")
plt.xlabel('test points')
plt.ylabel('prices')
plt.grid(True)
```
